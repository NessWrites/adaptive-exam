# Calculating discrimination

## 1. Simplified approach using point biserial correlation
reference: https://www.ijstr.org/final-print/apr2020/Test-Item-Analyzer-Using-Point-biserial-Correlation-And-P-values.pdf


### Point-Biserial Correlation ($r_{pb}$) Explained

The **point-biserial correlation ($r_{pb}$)** is a statistical measure ranging from -1.0 to +1.0. It quantifies the strength and direction of the relationship between a **binary variable** (e.g., a correct/incorrect response to a single test item) and a **continuous variable** (e.g., a student's total score on the entire test).

***

### What It Tells Us: The "Why"

The purpose of this correlation is to determine if a test item is effective at **discriminating** between high-ability students and low-ability students. A good question should be answered correctly more often by those who perform well on the overall test.

* **A high, positive $r_{pb}$** (e.g., > 0.3) is **good**. 
    This indicates that students who answered the item correctly tended to have high total scores, while those who answered it incorrectly had low total scores. The item successfully differentiates between knowledgeable and less knowledgeable students.

* **An $r_{pb}$ near zero** is **bad**. 
    This means there is no relationship between getting the item right and the total test score. High-scoring students are just as likely to get the question wrong as low-scoring students. Such an item does not contribute to the test's ability to discriminate.

* **A negative $r_{pb}$** is **very bad**. 
    This is a major red flag, indicating that students who answered the item correctly tended to have *lower* total scores. This can be caused by a mis-keyed answer, an ambiguous question, or a significant flaw that leads stronger students to choose the wrong option. 

***

### A Demonstration: The "How"

Let's use **Item 3** from your data, which has the strongest correlation ($r_{pb}$ = 0.768), to illustrate the concept.

| Item | $n_1$ (Correct) | $n_0$ (Incorrect) | Mean Score (Correct) | Mean Score (Incorrect) | $r_{pb}$ |
|:---|:---:|:---:|:---:|:---:|:---:|
| **Item 3** | 80 | 20 | 7.8 | 3.0 | 0.768 |

1.  **The Binary Variable**: For each student, we have a binary outcome for Item 3: they either got it right (coded as 1) or wrong (coded as 0).
    * 80 students got it right ($n_1$).
    * 20 students got it wrong ($n_0$).

2.  **The Continuous Variable**: Each student's total score on the entire test.

3.  **The Comparison**:
    * The **Mean Score (Correct)** is 7.8. This is the average total test score for the 80 students who correctly answered Item 3.
    * The **Mean Score (Incorrect)** is 3.0. This is the average total test score for the 20 students who got Item 3 wrong.

4.  **The "Aha!" Moment**:
    There is a significant gap of **4.8 points** (7.8 - 3.0) between the average total scores of the two groups. This demonstrates that students who correctly answered this item were, on average, much stronger students than those who didn't. The **$r_{pb}$ coefficient of 0.768** is simply a precise mathematical value that quantifies this powerful relationship, taking into account both the size of the score gap and the distribution of scores.

## Demonstration: Calculating the Discrimination Parameter ($a$) in IRT


This demonstration shows how to calculate the **discrimination parameter** ($a$) in **Item Response Theory (IRT)** using a **point-biserial correlation ($r_{pb}$)** as a proxy. While this method is an approximation and precise IRT estimation requires specialized software and full response data, it provides a useful way to understand the concept.

***

### Assumptions for Demonstration

* We'll use **3 items** and students' total scores (as a proxy for their ability, $\theta$).
* The **point-biserial correlation ($r_{pb}$)** is calculated between item responses (coded as 1 for correct, 0 for incorrect) and total scores.
* We will use the following approximation formula to compute $a$:

$$a \approx \frac{1.7 \, r_{pb}}{\sqrt{1 - r_{pb}^2}}$$

***

### Hypothetical Data Setup

* **Total score** is the sum of correct responses across all items (maximum score = 10).
* For each item, we'll use the following variables:
    * $n_1$: Number of students who answered correctly.
    * $n_0$: Number of students who answered incorrectly.
    * $\bar{X}_1$: Mean total score of students who answered correctly.
    * $\bar{X}_0$: Mean total score of students who answered incorrectly.
    * $\sigma_X$: Standard deviation of total scores (assumed to be 2.5).
    * The total number of students is $n = n_1 + n_0 = 100$.
* The formula for the point-biserial correlation is:

$$
r_{pb} = \frac{\bar{X}_1 - \bar{X}_0}{\sigma_X} 
\sqrt{\frac{n_1 n_0}{n^2}}
$$

***

### Sample Calculations

#### Item 1
* $n_1 = 70$, $n_0 = 30$
* $\bar{X}_1 = 8.0$, $\bar{X}_0 = 5.0$
* **Point-Biserial Correlation:**
    $$
    r_{pb} = \frac{8.0 - 5.0}{2.5} \sqrt{\frac{70 \cdot 30}{100^2}} 
    = 0.55
    $$
* **Discrimination Parameter ($a$):**
    $$
    a = \frac{1.7 \cdot 0.55}{\sqrt{1 - 0.55^2}} \approx 1.12
    $$

***

#### Item 2
* $n_1 = 50$, $n_0 = 50$
* $\bar{X}_1 = 7.5$, $\bar{X}_0 = 4.5$
* **Point-Biserial Correlation:**
    $$
    r_{pb} = \frac{7.5 - 4.5}{2.5} \sqrt{\frac{50 \cdot 50}{100^2}} 
    = 0.60
    $$
* **Discrimination Parameter ($a$):**
    $$
    a = \frac{1.7 \cdot 0.6}{\sqrt{1 - 0.6^2}} \approx 1.28
    $$

***

#### Item 3
* $n_1 = 80$, $n_0 = 20$
* $\bar{X}_1 = 7.8$, $\bar{X}_0 = 3.0$
* **Point-Biserial Correlation:**
    $$
    r_{pb} = \frac{7.8 - 3.0}{2.5} \sqrt{\frac{80 \cdot 20}{100^2}} 
    = 0.768
    $$
* **Discrimination Parameter ($a$):**
    $$
    a = \frac{1.7 \cdot 0.768}{\sqrt{1 - 0.768^2}} \approx 2.04
    $$
