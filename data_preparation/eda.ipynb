```python
import pandas as pd
import numpy as np
```

```python
df1= pd.read_csv("exports/AIF 2025 Entrance Exam - April 5th.csv")
df2= pd.read_csv("exports/AIF 2025 Entrance Exam - April 6th.csv")
df3= pd.read_csv("exports/AIF 2025 Entrance Exam - April 7th.csv")
df4= pd.read_csv("exports/AIF 2025 Entrance Re Exam - April 8th.csv")
df5= pd.read_csv("exports/AIF 2025 Entrance Re Exam - April 9th.csv")
df6= pd.read_csv("exports/AIF 2025 Entrance Exam Sazience.csv")
```

```python
df1.shape
```

```python
# concat dataframe
df= pd.concat([df1, df2, df3, df4, df5, df6], ignore_index=True)
```

```python
df.head()
```

```python
df.shape
```

```python
# get date only
df['SUBMITTED DATE']= pd.to_datetime(df1['SUBMITTED DATE']).dt.date
```

```python
df1.head()
```

```python
df.drop(columns=['Violation Status', 'STATUS', 'Unnamed: 2', 'Unnamed: 6'], inplace=True)
```

```python
df.head()
```

```python
df["Email"] = df["Name"].str.extract(r'([a-z0-9._%+-]+@[a-z0-9.-]+\.[a-z]{2,})')

# Extract name (uppercase part before email)
df["Name_clean"] = df["Name"].str.extract(r'([A-Z ]+)(?=[a-z0-9._%+-]+@)')

# Strip extra spaces
df["Name_clean"] = df["Name_clean"].str.strip()
```

```python
df.head()
```

```python
df.drop_duplicates(inplace=True)
```
