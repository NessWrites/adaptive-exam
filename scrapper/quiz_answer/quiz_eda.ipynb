```python
import pandas as pd
from pathlib import Path
import json
```

```python
path_quiz = Path("all_quiz_answers.json")

# Load the JSON file
with path_quiz.open("r", encoding="utf-8") as f:
    data = json.load(f)

# Count top-level keys
if isinstance(data, dict):
    count = len(data.keys())
    print("Number of main keys:", count)
else:
    print("JSON root is not an object (dict), it's a", type(data))
```
